```python
from sklearn.ensemble import RandomForestClassifier
import cv2
import pickle
import numpy as np
import os
import csv
import time
from datetime import datetime

# Initialize video capture
video = cv2.VideoCapture(0)

# Load the face detection classifier
facedetect = cv2.CascadeClassifier(r"C:\Users\Mahammad Iliyas\Downloads\opencv\haarcascade_frontalface_default.xml")

# Load the saved labels and face data
with open(r'C:\Users\Mahammad Iliyas\Downloads\opencv\data\names.pkl', 'rb') as f:
    LABELS = pickle.load(f)
with open(r'C:\Users\Mahammad Iliyas\Downloads\opencv\data\faces_data.pkl', 'rb') as f:
    FACES = pickle.load(f)

# Initialize the Random Forest classifier and fit the data
rf_clf = RandomForestClassifier(n_estimators=100)
rf_clf.fit(FACES, LABELS)

# Define column names for attendance
COL_NAMES = ['NAME', 'TIME']

while True:
    ret, frame = video.read()

    # Convert the frame to grayscale for face detection
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    faces = facedetect.detectMultiScale(gray, 1.3, 5)

    for (x, y, w, h) in faces:
        # Crop the face and resize it for prediction
        crop_img = frame[y:y+h, x:x+w, :]
        resized_img = cv2.resize(crop_img, (50, 50)).flatten().reshape(1, -1)
        output = rf_clf.predict(resized_img)

        # Get the current timestamp
        ts = time.time()
        date = datetime.fromtimestamp(ts).strftime("%d-%m-%Y")
        timestamp = datetime.fromtimestamp(ts).strftime("%H:%M:%S")
        filename = r"C:\Users\Mahammad Iliyas\Downloads\opencv\Attendance_" + date + ".csv"

        # Create CSV file if it doesn't exist
        if not os.path.isfile(filename):
            with open(filename, "w", newline="") as csvfile:
                writer = csv.writer(csvfile)
                writer.writerow(COL_NAMES)

        # Prepare the attendance data
        attendance = [str(output[0]), str(timestamp)]

        # Append attendance to the CSV file for the day
        with open(filename, "a", newline="") as csvfile:
            writer = csv.writer(csvfile)
            writer.writerow(attendance)

        # Draw rectangles around the face and add text
        cv2.rectangle(frame, (x, y), (x+w, y+h), (0, 0, 255), 1)
        cv2.rectangle(frame, (x, y-40), (x+w, y), (50, 50, 255), -1)
        cv2.putText(frame, str(output[0]), (x, y-15), cv2.FONT_HERSHEY_COMPLEX, 1, (255, 255, 255), 2)

    # Show the frame
    cv2.imshow("Frame", frame)

    # Capture key press
    k = cv2.waitKey(1)

    if k == ord('q'):  # 'q' key to quit the program
        break

# Release the video capture and close all OpenCV windows
video.release()
cv2.destroyAllWindows()
```

```python
from sklearn.ensemble import RandomForestClassifier
import cv2
import pickle
import numpy as np
import os
import csv
import time
from datetime import datetime

# Initialize video capture
video = cv2.VideoCapture(0)

# Load the face detection classifier
facedetect = cv2.CascadeClassifier(r"C:\Users\Mahammad Iliyas\Downloads\opencv\haarcascade_frontalface_default.xml")

# Load the saved labels and face data
with open(r'C:\Users\Mahammad Iliyas\Downloads\opencv\data\names.pkl', 'rb') as f:
    LABELS = pickle.load(f)
with open(r'C:\Users\Mahammad Iliyas\Downloads\opencv\data\faces_data.pkl', 'rb') as f:
    FACES = pickle.load(f)

# Initialize the Random Forest classifier and fit the data
rf_clf = RandomForestClassifier(n_estimators=100)
rf_clf.fit(FACES, LABELS)

# Define column names for attendance
COL_NAMES = ['NAME', 'TIME']

# Create a set to track attendance (recognized faces)
recognized_faces = set()

while True:
    ret, frame = video.read()

    # Convert the frame to grayscale for face detection
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    faces = facedetect.detectMultiScale(gray, 1.3, 5)

    for (x, y, w, h) in faces:
        # Crop the face and resize it for prediction
        crop_img = frame[y:y+h, x:x+w, :]
        resized_img = cv2.resize(crop_img, (50, 50)).flatten().reshape(1, -1)
        output = rf_clf.predict(resized_img)

        # Get the predicted name
        name = str(output[0])

        # Check if the face has already been recognized
        if name not in recognized_faces:
            # Add the recognized face to the set to avoid duplicate entries
            recognized_faces.add(name)

            # Get the current timestamp
            ts = time.time()
            date = datetime.fromtimestamp(ts).strftime("%d-%m-%Y")
            timestamp = datetime.fromtimestamp(ts).strftime("%H:%M:%S")
            filename = r"C:\Users\Mahammad Iliyas\Downloads\opencv\Attendance_" + date + ".csv"

            # Create CSV file if it doesn't exist
            if not os.path.isfile(filename):
                with open(filename, "w", newline="") as csvfile:
                    writer = csv.writer(csvfile)
                    writer.writerow(COL_NAMES)

            # Prepare the attendance data
            attendance = [name, str(timestamp)]

            # Append attendance to the CSV file for the day
            with open(filename, "a", newline="") as csvfile:
                writer = csv.writer(csvfile)
                writer.writerow(attendance)

            # Draw rectangles around the face and add text
            cv2.rectangle(frame, (x, y), (x+w, y+h), (0, 0, 255), 1)
            cv2.rectangle(frame, (x, y-40), (x+w, y), (50, 50, 255), -1)
            cv2.putText(frame, name, (x, y-15), cv2.FONT_HERSHEY_COMPLEX, 1, (255, 255, 255), 2)

        else:
            # If face is already recognized, draw rectangle and label without recording attendance
            cv2.rectangle(frame, (x, y), (x+w, y+h), (0, 255, 0), 1)  # Green for already recorded
            cv2.putText(frame, f'{name} (Recorded)', (x, y-15), cv2.FONT_HERSHEY_COMPLEX, 1, (0, 255, 0), 2)

    # Show the frame
    cv2.imshow("Frame", frame)

    # Capture key press
    k = cv2.waitKey(1)

    if k == ord('q'):  # 'q' key to quit the program
        break

# Release the video capture and close all OpenCV windows
video.release()
cv2.destroyAllWindows()
```
